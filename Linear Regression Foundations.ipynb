**Linear Regression Foundations: Data Cleaning, EDA, Outliers, and Assumptions.**

**1. The Workflow of a Linear Regression Project**

Before building a model, we need to understand the dataset.

A typical workflow looks like this:

Import Python libraries.

Read the dataset.

Understand the rows and columns.

Perform Exploratory Data Analysis (EDA).

Check for missing values.

Investigate outliers.

Examine relationships between variables.

Prepare the data.

Fit the regression model.

Check regression assumptions and model diagnostics.

Evaluate the model.

Important: We should not immediately start fitting a model. First, we need to understand what our data contains and whether there are quality problems that could affect our results.

**4. Exploratory Data Analysis (EDA)**

*What is EDA?*

Exploratory Data Analysis, usually called EDA, is the process of examining and summarizing a dataset before conducting further analysis or building a model.

In simple terms:

EDA means getting to know your data before asking it to answer questions.

We investigate:

What variables are in the dataset?

What do their distributions look like?

Are values missing?

Are there unusual observations?

Are variables related?

Are there patterns that could affect our regression model?

**Look at numerical distributions**

A histogram helps us understand how values are distributed.

For example:

df["income"].hist(bins=20)

plt.xlabel("Income")
plt.ylabel("Frequency")
plt.title("Distribution of Income")

plt.show()

**What does a histogram show?**

X-axis: Values of the variable.

Y-axis: Number of observations in each range.

Bars: How frequently values occur.

A histogram can help us identify:

Symmetric distributions.

Right-skewed distributions.

Left-skewed distributions.

Multiple peaks.

Possible unusual values.

**Check skewness**

df["income"].skew()

Skewness                  General interpretation

Around 0                  Approximately symmetric

Positive                  Right-skewed

Negative                  Left-skewed

A right-skewed income distribution might contain many ordinary incomes and a small number of extremely high incomes.

An outlier might be:

A measurement error.

A genuine unusual observation.

A member of a different population or group.

 **Find the minimum and maximum**

print("Minimum:", df["income"].min())

print("Maximum:", df["income"].max())

**What should we do with outliers?**

There is no single correct solution. We need to understand the observation first.


Keep      Retain a valid observation

Correct    Fix a confirmed data-entry or measurement error

Remove     Exclude an observation when there is a justified reason

Cap        Replace extreme values beyond chosen boundaries

Transform  Change the variable's mathematical scale

**Option 1: Keep the outlier**

If the observation is genuine and relevant to the population, keeping it may be appropriate.

Option 2: Remove an incorrect observation

If the value is clearly a data-entry error and cannot be corrected, removal may be justified.

Option 3: Cap extreme values

Capping is also called winsorization.

df["income_capped"] = df["income"].clip(
    lower=lower_bound,
    upper=upper_bound
)

This replaces values below the lower boundary with the lower boundary and values above the upper boundary with the upper boundary.

**Option 4: Apply a transformation**

For positive income values:

df["log_income"] = np.log(df["income"])

A logarithm compresses large values.

It may help with skewness or a nonlinear relationship, but it does not automatically fix every regression problem.

*Linear Regression Assumptions*

**what are regression assumptions?**

Regression assumptions are conditions about the data and the regression model that help us determine whether our estimates, standard errors, and statistical conclusions are reliable.

We do not simply fit a model and assume everything is fine.

We investigate the assumptions using plots, data knowledge, and diagnostic tests.

**Assumption 1: Linearity**

What does linearity mean?

Linearity means that the expected value of the outcome has a linear relationship with the predictors, given the other predictors in the model.

In simple terms, the model should be able to represent the relationship using a straight-line form.

**How do we check it before fitting the model?**

Use a scatterplot.

sns.scatterplot(
    data=df,
    x="age",
    y="income"
)

plt.title("Checking Linearity")
plt.show()

**What if we see a curved pattern?**

The relationship may not be adequately represented by a straight-line term.

One possibility is to add a squared predictor:

df["age_squared"] = df["age"] ** 2

Then a regression model can include both age and age squared.

**This is still a linear regression model in its coefficients, even though the relationship with age is curved.**

*How do we defend this assumption?*

**Assumption 2: Independence of errors**

What does independence mean?

The regression errors should not be systematically related across observations.

In simple terms, one observation's error should not provide information about another observation's error.

When might this become a problem?

Examples:

Repeated measurements from the same person.

Monthly sales observations.

Daily stock prices.

Students grouped within the same classroom.

These observations may be related to one another.

How do we investigate it?

First, understand how the data was collected.

Ask:

Are these observations from different independent individuals?

Are the same individuals measured repeatedly?

Is this time-series data?

Are observations grouped by location, school, company, or household?

For time-ordered data, plot residuals against time:

plt.plot(
    df["date"],
    residuals
)

plt.axhline(0, linestyle="--")

plt.title("Residuals Over Time")
plt.xlabel("Date")
plt.ylabel("Residuals")

plt.xticks(rotation=45)
plt.show()

Patterns or long runs of positive and negative residuals may suggest dependence.

**How do we defend this assumption?**

Explain how observations were collected and whether they are independent.

**If data is repeated or time-dependent, ordinary regression may need additional methods that account for the dependence.**

A graph alone cannot prove independence.

**Assumption 3: Homoscedasticity**

What does homoscedasticity mean?

Homoscedasticity means that the variance of the regression errors is approximately constant across the range of fitted values.

In simple terms:

The spread of the residuals should remain reasonably consistent.

How do we check it?

After fitting a regression model, calculate the residuals:

residuals = y_test - y_pred

Create a residual plot:

plt.scatter(
    y_pred,
    residuals
)

plt.axhline(
    y=0,
    color="red",
    linestyle="--"
)

plt.xlabel("Predicted Values")
plt.ylabel("Residuals")

plt.title("Residual Plot")

plt.show()

How do we read the plot?

A desirable pattern:

Dots scattered around zero.

No obvious curve.

No obvious funnel.

A reasonably consistent vertical spread.

A potential problem:

Residuals spread out as predictions increase.

Residuals become narrower as predictions increase.

A funnel or cone-shaped pattern appears.

These patterns may indicate heteroscedasticity.

What can we do if we find it?

Depending on the situation, possible approaches include:

Transforming the outcome.

**Using heteroscedasticity-robust standard errors for inference.**

Investigating whether the model is missing important predictors.

Using an appropriate weighted regression method.

**How do we defend this assumption?**

Show the residual-versus-fitted plot and explain whether the residual spread appears approximately constant.

If a clear funnel appears, acknowledge the issue and explain how you addressed it.

**Assumption 4: Normality of errors**

What does normality mean?

For conventional small-sample t-tests and confidence intervals, the regression errors are commonly assumed to be approximately normally distributed.

**Assumption 5: No perfect multicollinearity**

What does multicollinearity mean?

Multicollinearity occurs when two or more independent variables are strongly related to one another.

For example:

Age and years of work experience.

House size and number of rooms.

Height measured in centimeters and height measured in inches.

Some relationships are expected, but perfect multicollinearity prevents us from separately estimating the effects of the affected predictors.

High multicollinearity can make coefficients unstable and difficult to interpret.

**How do we check it?**

Start with a correlation matrix:

correlation = df[
    ["age", "education", "experience"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Predictor Correlations")
plt.show()

Then investigate Variance Inflation Factor (VIF):

from statsmodels.stats.outliers_influence import variance_inflation_factor

X = df[
    ["age", "education", "experience"]
]

vif = pd.DataFrame()

vif["Variable"] = X.columns

vif["VIF"] = [
    variance_inflation_factor(
        X.values,
        i
    )
    for i in range(X.shape[1])
